# SECTION 1: Auto Loader Introduction

## What is Auto Loader?

**Auto Loader** is Databricks' **incremental data ingestion solution** that automatically detects and processes new files as they arrive in cloud storage (S3, ADLS, GCS). It's built on Apache Spark Structured Streaming and provides:

* **Incremental Processing**: Only new files are processed, not the entire directory
* **Automatic Schema Management**: Detects and adapts to schema changes automatically
* **Exactly-Once Guarantees**: Built-in checkpointing prevents duplicate processing
* **Scalability**: Efficiently handles millions of files
* **Cost Efficiency**: Reduces compute costs by 70-90% compared to full reprocessing

## Why Enterprises Use Auto Loader

### Business Impact
* **Cost Reduction**: 70-90% lower compute costs vs batch reprocessing
  * Example: Processing 10TB daily - Batch: $500/day, Auto Loader: $50-150/day
* **Reduced Latency**: Near real-time ingestion (seconds to minutes vs hours)
* **Operational Simplicity**: No manual file tracking or orchestration needed
* **Data Quality**: Automatic schema validation and rescued data column
* **Scalability**: Production systems handle 100K+ files/hour

### Technical Benefits
* Schema evolution without downtime
* Automatic retry and error handling
* Built-in exactly-once semantics
* No external dependencies (Kafka, message queues)

## Comparison: spark.read vs COPY INTO vs Auto Loader

| Feature | spark.read | COPY INTO | Auto Loader |
|---------|-----------|-----------|-------------|
| **Processing Model** | Full scan every time | Incremental (tracks ingested files) | Incremental (streaming) |
| **Cost for 1000 files** | Processes all 1000 files every run | Only new files | Only new files |
| **Schema Evolution** | Manual code changes required | Limited support | Automatic with configurable modes |
| **Exactly-Once** | No (manual deduplication) | Yes (idempotent) | Yes (checkpointing) |
| **File Discovery** | Directory listing | Directory listing | Directory listing OR file notification |
| **Scalability** | Poor (10K+ files = slow) | Good (tracks processed files) | Excellent (optimized for millions) |
| **Latency** | Batch only | Batch only | Near real-time |
| **Use Case** | One-time loads, ad-hoc analysis | Periodic batch ingestion | Continuous/streaming ingestion |
| **Production Ready** | No (for incremental) | Yes (for batch) | Yes (for streaming) |
| **Error Handling** | Manual | Good | Excellent (_rescued_data) |
| **Cost (relative)** | 100% (baseline) | 20-30% | 10-20% |

**Interview Insight**: COPY INTO is SQL-native and great for batch ETL pipelines. Auto Loader is better for streaming/near-realtime and complex schema evolution scenarios.

## Incremental Ingestion Concept

### Traditional Approach (spark.read)
```python
# Day 1: Process 100 files
df = spark.read.parquet("/data/")
# Processes all 100 files

# Day 2: 10 new files added, now 110 total
df = spark.read.parquet("/data/")
# Processes ALL 110 files again! (100 old + 10 new)
```
**Problem**: Wasted compute on already-processed data.

### Auto Loader Approach
```python
# Day 1: Process 100 files
df = spark.readStream.format("cloudFiles") \
    .option("cloudFiles.format", "parquet") \
    .load("/data/")
# Processes 100 files, saves checkpoint

# Day 2: 10 new files added
# Auto Loader reads checkpoint, processes ONLY the 10 new files!
```
**Benefit**: 90% cost reduction in this example.

## File Discovery Mechanisms

Auto Loader offers two modes for detecting new files:

### 1. Directory Listing (Default)
* **How it works**: Periodically lists the directory to find new files
* **Best for**: 
  * < 10,000 files in directory
  * Infrequent arrivals (every few minutes/hours)
  * Simple cloud storage setups
* **Cost**: Cloud API calls for listing (typically $0.005 per 1000 requests)
* **Latency**: Configurable (default: every trigger)
* **No setup required**

### 2. File Notification (Advanced)
* **How it works**: Uses cloud-native notifications (S3 events, Azure Event Grid, GCS Pub/Sub)
* **Best for**:
  * > 10,000 files in directory
  * High-frequency arrivals (many files per second)
  * Large-scale production systems
* **Cost**: Notification service costs (typically $0.50 per million notifications)
* **Latency**: Sub-second detection
* **Requires**: Cloud notification setup (event subscriptions)

**Configuration**:
```python
.option("cloudFiles.useNotifications", "true")  # Enable file notifications
```

**Interview Question**: "When would you choose file notification over directory listing?"
**Answer**: When you have 10K+ files or need sub-second latency. Trade-off is additional cloud infrastructure setup.

## Common Misconceptions

### ❌ Myth 1: "Auto Loader only works for streaming"
**Reality**: Works for both streaming and batch. Use `.trigger(availableNow=True)` for batch mode with incremental benefits.

### ❌ Myth 2: "Checkpoint can never be deleted"
**Reality**: You CAN delete checkpoints for full replay, but understand the implications (reprocessing all historical data).

### ❌ Myth 3: "Schema evolution means any schema change is OK"
**Reality**: Type changes (string → int) can break downstream queries. Use schema evolution modes carefully.

### ❌ Myth 4: "Auto Loader is only for raw/bronze layer"
**Reality**: While common in bronze, can be used anywhere files arrive (external vendor drops, ML model outputs).

### ❌ Myth 5: "Works only with Delta tables"
**Reality**: Can write to any Spark-supported format (Delta, Parquet, Iceberg), but Delta gives you ACID guarantees.

The key architectural components are: 
1. **File discovery** - either directory listing or cloud-native file notifications for scale

````python
# PRODUCTION AUTO LOADER EXAMPLE - JSON TO BRONZE
# Complete end-to-end example using Unity Catalog best practices

from pyspark.sql.functions import current_timestamp, to_date, col

print("=" * 80)
print("PRODUCTION AUTO LOADER: JSON TO BRONZE LAYER")
print("=" * 80)

# Get configuration for JSON IoT pipeline
json_config = configs["json_iot"]

print(f"\n📁 Pipeline Configuration:")
print(f"   Source: {json_config['source']}")
print(f"   Format: {json_config['format']}")
print(f"   Checkpoint: {json_config['checkpoint']}")
print(f"   Schema Location: {json_config['schema_location']}")
print(f"   Target Table: {json_config['target_table']}")
print()

# Configure Auto Loader stream
print("🔧 Configuring Auto Loader stream...\n")

raw_stream = (spark.readStream
    .format("cloudFiles")  # Auto Loader format
    
    # Source format
    .option("cloudFiles.format", json_config["format"])
    
    # Schema management (production settings)
    .option("cloudFiles.schemaLocation", json_config["schema_location"])
    .option("cloudFiles.inferColumnTypes", "true")
    .option("cloudFiles.schemaEvolutionMode", "rescue")  # Production: use rescue mode
    
    # Performance tuning
    .option("cloudFiles.maxFilesPerTrigger", 1000)  # Batch size
    
    # Load from source
    .load(json_config["source"])
)

# Add bronze layer metadata columns (production standard)
bronze_stream = (raw_stream
    .withColumn("_ingestion_timestamp", current_timestamp())
    .withColumn("_source_file", col("_metadata.file_path"))
    .withColumn("_load_date", to_date(current_timestamp()))
)

print("✅ Stream configured with metadata columns:\n")
bronze_stream.printSchema()

print("\n" + "-" * 80)
print("👁️  PRODUCTION FEATURES ENABLED:")
print("-" * 80)
print("""
1. ✅ Unity Catalog volumes for checkpoints/schemas
2. ✅ Schema evolution in rescue mode (production-safe)
3. ✅ Metadata columns for auditability:
   - _ingestion_timestamp: When data arrived
   - _source_file: Which file it came from
   - _load_date: Partition key for efficient queries
4. ✅ Batch size control (maxFilesPerTrigger=1000)
5. ✅ Target: Bronze layer in Unity Catalog
6. ✅ _rescued_data column (automatic) for malformed records
""")

print("-" * 80)
print("🚀 Ready to write to bronze layer!")
print(f"   Target: {json_config['target_table']}")
print(f"   Partition: _load_date")
print("-" * 80)

print("\n📝 To execute this stream, run:")
print("""
(bronze_stream.writeStream
    .format("delta")
    .option("checkpointLocation", json_config["checkpoint"])
    .partitionBy("_load_date")
    .option("mergeSchema", "true")
    .trigger(availableNow=True)  # Batch mode
    .table(json_config["target_table"]))
""")

print("=" * 80)
print("✅ Production Auto Loader configuration complete!")
print("=" * 80)
````

# SECTION 3: JSON Ingestion with Auto Loader

## Overview

JSON is the most common format for:
* REST API responses
* Application logs
* IoT device telemetry
* Event streams

**Challenge**: JSON files often have:
* Nested structures
* Varying schemas (different devices send different fields)
* Schema evolution over time

**Auto Loader Solution**: Automatically handles all these challenges!

## Key Auto Loader Parameters for JSON

Let's understand each parameter before implementation:

### 1. `cloudFiles.format`
* **What**: Specifies the source file format
* **Values**: `json`, `csv`, `parquet`, `avro`, `orc`, `text`, `binaryFile`
* **Production Note**: Always specify explicitly - never rely on auto-detection
* **Interview Point**: "This tells Auto Loader how to parse the raw bytes"

### 2. `cloudFiles.schemaLocation`
* **What**: Directory where Auto Loader stores inferred schema metadata
* **Why Critical**: 
  * Enables schema evolution tracking
  * Prevents schema inference on every run (performance)
  * Acts as schema "memory" across restarts
* **Best Practice**: Use a dedicated location per pipeline
* **Production Note**: Should be in a persistent, backed-up location
* **Interview Point**: "This is where Auto Loader remembers what schema it has seen"

### 3. `cloudFiles.inferColumnTypes`
* **What**: Whether to infer data types or read everything as strings
* **Values**: `true` (infer types) or `false` (all strings)
* **Default**: `false`
* **Production Recommendation**: 
  * `true` for structured data (IoT, application logs)
  * `false` for unpredictable data (scraping, external vendors)
* **Trade-off**: Type safety vs flexibility
* **Interview Point**: "False is safer but requires downstream type casting"

### 4. `cloudFiles.schemaEvolutionMode`
* **What**: How to handle schema changes (new columns, type changes)
* **Values**:
  * `addNewColumns` - Allow new columns (default)
  * `rescue` - Route unexpected data to _rescued_data column
  * `failOnNewColumns` - Fail the stream if schema changes
  * `none` - No schema evolution
* **Production Guidance**:
  * Bronze layer: `addNewColumns` (capture everything)
  * Silver/Gold: `rescue` or `failOnNewColumns` (controlled)
* **Interview Point**: "This controls the trade-off between flexibility and strict schema enforcement"

````python
# PRODUCTION-READY AUTO LOADER FOR JSON
# This example demonstrates all key parameters with inline explanations

from pyspark.sql.functions import current_timestamp, input_file_name

# Configuration
source_path = "/databricks-datasets/iot-stream/data-device/"  # Source JSON files
checkpoint_path = "/tmp/autoloader_checkpoints/json_iot"      # Checkpoint location
schema_location = "/tmp/autoloader_schema/json_iot"           # Schema inference location
target_table = "json_iot_raw"                                  # Target Delta table name

print("=" * 80)
print("AUTO LOADER: JSON INGESTION DEMO")
print("=" * 80)
print(f"\nSource Path: {source_path}")
print(f"Checkpoint: {checkpoint_path}")
print(f"Schema Location: {schema_location}")
print(f"Target Table: {target_table}")
print("\n" + "=" * 80)

# Auto Loader stream configuration
df = (spark.readStream
    .format("cloudFiles")  # This is the Auto Loader format!
    
    # Parameter 1: Specify source file format
    .option("cloudFiles.format", "json")
    # Why: Tells Auto Loader these are JSON files
    # Production: Always explicit, never auto-detect
    
    # Parameter 2: Schema tracking location
    .option("cloudFiles.schemaLocation", schema_location)
    # Why: Stores inferred schema, enables evolution tracking
    # Production: Use persistent storage, one location per pipeline
    # Interview: "This is Auto Loader's schema memory"
    
    # Parameter 3: Infer column types (vs all strings)
    .option("cloudFiles.inferColumnTypes", "true")
    # Why: Automatically infers int, double, timestamp, etc.
    # Production: true for structured data, false for unpredictable data
    # Trade-off: Type safety vs flexibility
    
    # Parameter 4: Schema evolution strategy
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
    # Why: Automatically adds new columns as they appear
    # Production: 
    #   - Bronze layer: "addNewColumns" (capture everything)
    #   - Silver/Gold: "rescue" or "failOnNewColumns" (controlled)
    # Interview: "Balances flexibility vs strict schema enforcement"
    
    # Parameter 5: Include metadata column for rescued data
    .option("cloudFiles.useIncrementalListing", "auto")
    # Why: Optimizes file discovery (directory listing vs notifications)
    # Production: "auto" lets Databricks choose based on file count
    
    .load(source_path)  # Load from source path
)

# Add audit metadata (production best practice)
df_with_metadata = df \
    .withColumn("ingestion_timestamp", current_timestamp()) \
    .withColumn("source_file", input_file_name())

print("\n✅ Auto Loader stream configured successfully!")
print("\n📈 Inferred Schema:")
df_with_metadata.printSchema()

print("\n" + "=" * 80)
print("INTERVIEW TALKING POINTS")
print("=" * 80)
print("""
1. cloudFiles.format="json" - Tells Auto Loader the file type
2. schemaLocation - Where Auto Loader stores schema metadata
3. inferColumnTypes=true - Infer data types (vs all strings)
4. schemaEvolutionMode="addNewColumns" - Allow new fields automatically
5. Added metadata columns for auditability (ingestion_timestamp, source_file)

Key Interview Point:
"Auto Loader provides exactly-once processing through checkpointing,
automatically handles schema evolution, and adds negligible overhead
compared to manual file tracking solutions."
""")
print("=" * 80)
````

# SECTION 4: CSV Ingestion with Auto Loader

## CSV-Specific Challenges

CSV files are ubiquitous but challenging:
* **No built-in schema** - must infer or provide explicitly
* **Delimiter variations** - comma, pipe, tab, semicolon
* **Header handling** - first row may or may not be headers
* **Data type ambiguity** - everything looks like a string
* **Quote/escape handling** - values containing delimiters

## Two Approaches

### Approach 1: Schema Inference (Easier, Less Control)
* Let Auto Loader figure out the schema
* **Pros**: Quick to implement, adapts to changes
* **Cons**: May infer wrong types, performance overhead on first run
* **Use When**: Exploratory work, trusted data sources

### Approach 2: Explicit Schema (More Control, Production Standard)
* Define schema upfront using StructType
* **Pros**: Predictable types, better performance, catches errors early
* **Cons**: More code, must update when schema changes
* **Use When**: Production pipelines, compliance requirements

## Key CSV Parameters

* `header`: `true` if first row is column names, `false` otherwise
* `inferSchema`: Let Spark guess types (slower, use sparingly)
* `delimiter`: Character separating fields (default: comma)
* `quote`: Character for quoting values (default: double-quote)
* `escape`: Character for escaping special chars (default: backslash)
* `multiLine`: `true` if values can span multiple lines
* `encoding`: Character encoding (UTF-8, ISO-8859-1, etc.)

````python
# APPROACH 1: AUTO LOADER CSV WITH SCHEMA INFERENCE
# Simpler but less control - good for exploration

from pyspark.sql.functions import current_timestamp, col

print("=" * 80)
print("CSV AUTO LOADER: SCHEMA INFERENCE APPROACH")
print("=" * 80)

# Configuration
csv_source = "/databricks-datasets/bikeSharing/data-001/"
csv_checkpoint = "/Volumes/poc_deepdive/_infrastructure/checkpoints/csv_inference"
csv_schema_loc = "/Volumes/poc_deepdive/_infrastructure/schemas/csv_inference"

print(f"\nSource: {csv_source}")
print(f"Approach: Schema Inference (Auto Loader figures it out)")

# Auto Loader with schema inference
csv_df_inferred = (spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "csv")  # Specify CSV format
    
    # CSV-specific options
    .option("header", "true")  # First row contains column names
    .option("inferSchema", "true")  # Let Auto Loader infer data types
    # Note: This adds overhead on first run as Auto Loader samples the data
    
    .option("cloudFiles.schemaLocation", csv_schema_loc)
    .option("cloudFiles.schemaEvolutionMode", "addNewColumns")
    
    .load(csv_source)
)

print("\n📈 Inferred Schema:")
csv_df_inferred.printSchema()

print("\n" + "-" * 80)
print("👍 PROS OF SCHEMA INFERENCE")
print("-" * 80)
print("""
1. Quick implementation - minimal code
2. Adapts automatically if source schema changes
3. Good for exploratory data analysis
4. No need to know schema upfront
""")

print("👎 CONS OF SCHEMA INFERENCE")
print("-" * 80)
print("""
1. Performance overhead on first run (samples data)
2. May infer wrong types (e.g., ZIP codes as integers)
3. Less predictable in production
4. Harder to catch schema issues early
""")

print("\n🎯 WHEN TO USE: Exploratory work, trusted data sources\n")
print("=" * 80)
````

# SECTION 5: Parquet Ingestion with Auto Loader

## Why Parquet is Different

Parquet files are **self-describing** - the schema is embedded in the file format itself. This makes Auto Loader ingestion simpler:

* **No schema inference needed** - schema is already there
* **Type safety guaranteed** - integers stay integers, decimals stay decimals
* **Better performance** - columnar format, built-in compression
* **Production standard** - most data lakes use Parquet

## Parquet vs CSV/JSON

| Feature | JSON/CSV | Parquet |
|---------|----------|----------|
| Schema included? | No - must infer | Yes - embedded |
| inferSchema option? | Needed | Not needed |
| Read performance | Slower | Faster (columnar) |
| Storage efficiency | Poor | Excellent (compression) |
| Type preservation | Lost (all strings) | Preserved |
| Production ready? | Needs transformation | Yes |

## Enterprise Use Cases

1. **Data Warehouse Unloads**
   * Snowflake, Redshift export to S3 as Parquet
   * Auto Loader ingests directly to Delta
   
2. **Spark Job Outputs**
   * One Spark job writes Parquet
   * Auto Loader picks up for next stage
   
3. **ML Feature Stores**
   * Features computed and saved as Parquet
   * Training pipelines use Auto Loader

4. **Analytics Datasets**
   * Pre-aggregated, optimized for queries
   * Auto Loader for incremental updates

# DATA QUALITY FRAMEWORK
from pyspark.sql.functions import col, when, count, lit

class DataQualityCheck:
    def __init__(self, df, table_name):
        self.df = df
        self.table_name = table_name
        self.total_records = df.count()
        
    def check_rescued_data(self, threshold=0.01):
        """Alert if >1% of records in rescued_data"""
        rescued_count = self.df.filter(col("_rescued_data").isNotNull()).count()
        rescued_pct = rescued_count / self.total_records if self.total_records > 0 else 0
        
        if rescued_pct > threshold:
            raise ValueError(
                f"🚨 DATA QUALITY ALERT: {rescued_pct*100:.2f}% of records "
                f"in _rescued_data (threshold: {threshold*100}%). "
                f"Check source data format!"
            )
        print(f"✅ Rescued data: {rescued_pct*100:.4f}% (OK)")
        
    def check_nulls(self, columns, threshold=0.1):
        """Alert if critical columns have >10% nulls"""
        for col_name in columns:
            null_count = self.df.filter(col(col_name).isNull()).count()
            null_pct = null_count / self.total_records
            
            if null_pct > threshold:
                raise ValueError(
                    f"🚨 NULL ALERT: Column '{col_name}' is "
                    f"{null_pct*100:.2f}% null (threshold: {threshold*100}%)"
                )
        print(f"✅ Null checks passed for {columns}")
        
    def check_value_range(self, column, min_val, max_val):
        """Validate numeric ranges"""
        out_of_range = self.df.filter(
            (col(column) < min_val) | (col(column) > max_val)
        ).count()
        
        if out_of_range > 0:

            # Log outliers to quarantine table
            self.df.filter(
                (col(column) < min_val) | (col(column) > max_val)
            ).write.mode("append").saveAsTable(f"{self.table_name}_quarantine")
        else:
            print(f"✅ Value range check passed for {column}")

# Pattern 1: Selective Date Range Replay
def backfill_date_range(start_date, end_date, source_table, target_table, checkpoint):
    """
    Reprocess specific date range without affecting other data.
    """
    print(f"Backfilling {start_date} to {end_date}...")
    
    # Step 1: Delete target data for date range
    spark.sql(f"""
        DELETE FROM {target_table}
        WHERE _load_date BETWEEN '{start_date}' AND '{end_date}'
    """)
    
    # Step 2: Create filtered source stream
    stream = (

        .table(source_table)
        .filter(f"_load_date BETWEEN '{start_date}' AND '{end_date}'")
    )
    
    # Step 3: Use NEW checkpoint (don't reuse production checkpoint!)
    backfill_checkpoint = f"{checkpoint}_backfill_{start_date}_{end_date}"
    
    query = (stream.writeStream
        .option("checkpointLocation", backfill_checkpoint)
        .trigger(availableNow=True)
        .toTable(target_table))
    
    query.awaitTermination()
    print(f"✅ Backfill complete. Cleaning up checkpoint...")
    dbutils.fs.rm(backfill_checkpoint, recurse=True)

````python
# HANDS-ON EXERCISE 1: Complete Auto Loader Pipeline with Write
# This demonstrates the COMPLETE flow that was missing in earlier examples

from pyspark.sql.functions import current_timestamp, col, to_date

print("=" * 80)
print("HANDS-ON 1: END-TO-END AUTO LOADER PIPELINE")
print("=" * 80)

# Step 1: Configure paths
source_path = "/databricks-datasets/iot-stream/data-device/"
checkpoint_path = "/Volumes/poc_deepdive/_infrastructure/checkpoints/hands_on_iot"
schema_path = "/Volumes/poc_deepdive/_infrastructure/schemas/hands_on_iot"
target_table = "poc_deepdive.bronze.iot_telemetry_complete"

print(f"\n📂 Configuration:")
print(f"   Source: {source_path}")
print(f"   Checkpoint: {checkpoint_path}")
print(f"   Schema: {schema_path}")
print(f"   Target: {target_table}")

# Step 2: Clean up previous runs (for repeatability)
print("\n🧹 Cleaning up previous runs...")
try:
    spark.sql(f"DROP TABLE IF EXISTS {target_table}")
    dbutils.fs.rm(checkpoint_path, recurse=True)
    dbutils.fs.rm(schema_path, recurse=True)
    print("✅ Cleanup complete")
except:
    print("✅ No previous data to clean")

# Step 3: Configure Auto Loader stream
print("\n🔧 Configuring Auto Loader stream...")
stream = (spark.readStream
    .format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", schema_path)
    .option("cloudFiles.inferColumnTypes", "true")
    .option("cloudFiles.schemaEvolutionMode", "rescue")
    .option("cloudFiles.maxFilesPerTrigger", 10)  # Process 10 files per batch
    .load(source_path)
)

# Step 4: Add metadata columns (production standard)
print("\n📊 Adding metadata columns...")
stream_with_metadata = (stream
    .withColumn("_ingestion_timestamp", current_timestamp())
    .withColumn("_source_file", col("_metadata.file_path"))
    .withColumn("_load_date", to_date(current_timestamp()))
)

print("✅ Stream configured")
print("\nSchema:")
stream_with_metadata.printSchema()

# Step 5: Write to Delta table (THE CRITICAL PART!)
print("\n" + "=" * 80)
print("🚀 WRITING TO DELTA TABLE (This is what was missing!)")
print("=" * 80)

query = (stream_with_metadata.writeStream
    .format("delta")
    .option("checkpointLocation", checkpoint_path)
    .option("mergeSchema", "true")  # Allow schema evolution
    .partitionBy("_load_date")  # Partition for efficient queries
    .trigger(availableNow=True)  # Process all available files then stop
    .toTable(target_table)
)

print("⏳ Processing files... (this may take 30-60 seconds)")
query.awaitTermination()

print("\n✅ Write complete!")

# Step 6: Verify results
print("\n" + "=" * 80)
print("📊 VERIFICATION")
print("=" * 80)

df = spark.table(target_table)
total_count = df.count()
print(f"\n✅ Total records ingested: {total_count:,}")

# Check rescued data
rescued_count = df.filter(col("_rescued_data").isNotNull()).count()
rescued_pct = (rescued_count / total_count * 100) if total_count > 0 else 0
print(f"📋 Rescued data records: {rescued_count} ({rescued_pct:.2f}%)")

# Sample data
print("\n📄 Sample records:")
df.select(
    "device_id", "calories_burnt", "num_steps", 
    "_ingestion_timestamp", "_load_date"
).show(5, truncate=False)

# Check partitions
print("\n🗂️  Partitions created:")
spark.sql(f"SHOW PARTITIONS {target_table}").show()

print("\n" + "=" * 80)
print("✅ COMPLETE! You now have a working Auto Loader pipeline.")
print("=" * 80)
print("\n💡 Key Takeaways:")
print("   1. writeStream.toTable() is REQUIRED to actually write data")
print("   2. trigger(availableNow=True) = process then stop (cost-efficient)")
print("   3. Checkpoint enables exactly-once processing")
print("   4. Metadata columns (_ingestion_timestamp, _source_file) enable auditability")
print("   5. Partitioning by _load_date enables efficient time-based queries")
````

````python
# HANDS-ON EXERCISE 3: Complete Medallion Architecture
# Bronze → Silver → Gold (all streaming!)

from pyspark.sql.functions import (
    col, to_timestamp, to_date, when, avg, sum, count,
    current_timestamp, window
)

print("=" * 80)
print("HANDS-ON 3: COMPLETE MEDALLION PIPELINE")
print("=" * 80)
print("\n🎯 Goal: Build Bronze → Silver → Gold (all incremental)")

# ============================================================================
# LAYER 1: BRONZE (Already created in Exercise 1)
# ============================================================================
print("\n" + "=" * 80)
print("📦 BRONZE LAYER: Raw Ingestion (already complete)")
print("=" * 80)

bronze_table = "poc_deepdive.bronze.iot_telemetry_complete"
bronze_count = spark.table(bronze_table).count()
print(f"✅ Bronze records: {bronze_count:,}")

# ============================================================================
# LAYER 2: SILVER (Cleansed & Conformed)
# ============================================================================
print("\n" + "=" * 80)
print("🥈 SILVER LAYER: Cleansing & Transformation")
print("=" * 80)

silver_table = "poc_deepdive.silver.iot_telemetry_cleaned"
silver_checkpoint = "/Volumes/poc_deepdive/_infrastructure/checkpoints/silver_iot"

# Clean up previous runs
print("\n🧹 Cleaning previous silver data...")
try:
    spark.sql(f"DROP TABLE IF EXISTS {silver_table}")
    dbutils.fs.rm(silver_checkpoint, recurse=True)
    print("✅ Cleanup complete")
except:
    print("✅ No previous data to clean")

print("\n🔧 Configuring Silver stream (reading from Bronze)...")

# Read from bronze as a STREAM (not batch!)
silver_stream = (spark.readStream
    .table(bronze_table)
    
    # Data Quality: Remove invalid records
    .filter(col("device_id").isNotNull())
    .filter(col("calories_burnt") >= 0)
    .filter(col("num_steps") >= 0)
    .filter(col("miles_walked") >= 0)
    
    # Remove duplicates (same device_id + timestamp)
    .dropDuplicates(["device_id", "timestamp"])
    
    # Type conversions
    .withColumn("timestamp_parsed", to_timestamp(col("timestamp")))
    .withColumn("date", to_date(col("timestamp_parsed")))
    
    # Business logic: Calculate efficiency metrics
    .withColumn("calories_per_step", 
                when(col("num_steps") > 0, 
                     col("calories_burnt") / col("num_steps"))
                .otherwise(0))
    
    .withColumn("steps_per_mile",
                when(col("miles_walked") > 0,
                     col("num_steps") / col("miles_walked"))
                .otherwise(0))
    
    # Add silver metadata
    .withColumn("_silver_timestamp", current_timestamp())
    
    # Select final columns
    .select(
        "device_id",
        "user_id",
        "timestamp_parsed",
        "date",
        "calories_burnt",
        "num_steps",
        "miles_walked",
        "calories_per_step",
        "steps_per_mile",
        "_silver_timestamp",
        "_ingestion_timestamp"  # Keep bronze timestamp for lineage
    )
)

print("\n📊 Silver schema:")
silver_stream.printSchema()

print("\n🚀 Writing to Silver table...")
silver_query = (silver_stream.writeStream
    .format("delta")
    .option("checkpointLocation", silver_checkpoint)
    .option("mergeSchema", "false")  # Silver: strict schema
    .partitionBy("date")
    .trigger(availableNow=True)
    .toTable(silver_table)
)

print("⏳ Processing...")
silver_query.awaitTermination()

silver_count = spark.table(silver_table).count()
print(f"\n✅ Silver records: {silver_count:,}")
print(f"📉 Records filtered out: {bronze_count - silver_count:,}")

# Show sample
print("\n📄 Silver sample:")
spark.table(silver_table).select(
    "device_id", "date", "calories_burnt", "num_steps",
    "calories_per_step", "steps_per_mile"
).show(5, truncate=False)

# ============================================================================
# LAYER 3: GOLD (Business Aggregates)
# ============================================================================
print("\n" + "=" * 80)
print("🥇 GOLD LAYER: Business Aggregations")
print("=" * 80)

gold_table = "poc_deepdive.gold.iot_daily_summary"
gold_checkpoint = "/Volumes/poc_deepdive/_infrastructure/checkpoints/gold_iot"

# Clean up previous runs
print("\n🧹 Cleaning previous gold data...")
try:
    spark.sql(f"DROP TABLE IF EXISTS {gold_table}")
    dbutils.fs.rm(gold_checkpoint, recurse=True)
    print("✅ Cleanup complete")
except:
    print("✅ No previous data to clean")

print("\n🔧 Configuring Gold stream (reading from Silver)...")

# Read from silver as a STREAM
gold_stream = (spark.readStream
    .table(silver_table)
    
    # Aggregate by device per day
    .groupBy("device_id", "user_id", "date")
    .agg(
        count("*").alias("reading_count"),
        sum("num_steps").alias("total_steps"),
        sum("calories_burnt").alias("total_calories"),
        sum("miles_walked").alias("total_miles"),
        avg("calories_per_step").alias("avg_calories_per_step"),
        avg("steps_per_mile").alias("avg_steps_per_mile")
    )
    
    .withColumn("_gold_timestamp", current_timestamp())
)

print("\n📊 Gold schema:")
gold_stream.printSchema()

print("\n🚀 Writing to Gold table...")
gold_query = (gold_stream.writeStream
    .format("delta")
    .option("checkpointLocation", gold_checkpoint)
    .option("mergeSchema", "false")  # Gold: strict schema
    .outputMode("complete")  # Aggregations need complete mode
    .partitionBy("date")
    .trigger(availableNow=True)
    .toTable(gold_table)
)

print("⏳ Processing...")
gold_query.awaitTermination()

gold_count = spark.table(gold_table).count()
print(f"\n✅ Gold records: {gold_count:,}")

# Show sample
print("\n📄 Gold sample (Daily device summaries):")
spark.table(gold_table).select(
    "device_id", "date", "reading_count", "total_steps",
    "total_calories", "total_miles"
).orderBy("date", "device_id").show(10, truncate=False)

# ============================================================================
# SUMMARY
# ============================================================================
print("\n" + "=" * 80)
print("✅ MEDALLION PIPELINE COMPLETE!")
print("=" * 80)
print(f"\n📊 Pipeline Summary:")
print(f"   Bronze (raw):      {bronze_count:,} records")
print(f"   Silver (cleaned):  {silver_count:,} records ({silver_count/bronze_count*100:.1f}%)")
print(f"   Gold (aggregated): {gold_count:,} records (daily summaries)")

print("\n🎯 Key Achievements:")
print("   ✅ All three layers are STREAMING (incremental)")
print("   ✅ Bronze → Silver: Data quality filters + deduplication")
print("   ✅ Silver → Gold: Daily aggregations per device")
print("   ✅ Each layer has independent checkpoint")
print("   ✅ Adding 1GB to bronze only processes 1GB in silver/gold")

print("\n💡 Production Benefits:")
print("   • 90% cost reduction vs batch reprocessing")
print("   • Sub-second latency from bronze to gold")
print("   • Automatic recovery from failures")
print("   • Clear separation of concerns (raw/cleaned/business)")

print("\n🎤 Interview Answer:")
print("""
"We stream all three medallion layers. Bronze captures raw data via Auto Loader.
Silver reads bronze as a stream using spark.readStream.table() - this is key,
it makes downstream transformations incremental. We cleanse, deduplicate, and
calculate business metrics in silver. Gold aggregates by time window.
The entire pipeline is incremental - adding 1GB to bronze triggers only 1GB
processing in silver and gold, not full reprocessing. We use separate checkpoints
per layer. This architecture processes terabytes daily at 10% the cost of batch ETL."
""")
````

````python
# HANDS-ON EXERCISE 5: Data Quality Framework
# Production pipelines MUST validate data quality at each layer

from pyspark.sql.functions import col, count, when, sum as spark_sum

print("=" * 80)
print("HANDS-ON 5: DATA QUALITY VALIDATION FRAMEWORK")
print("=" * 80)
print("\n🎯 Goal: Build automated quality gates")

bronze_table = "poc_deepdive.bronze.iot_telemetry_complete"
df = spark.table(bronze_table)
total_records = df.count()

# ============================================================================
# QUALITY CHECK 1: Rescued Data Monitoring
# ============================================================================
print("\n" + "=" * 80)
print("🔍 CHECK 1: RESCUED DATA (Schema Issues)")
print("=" * 80)

rescued_records = df.filter(col("_rescued_data").isNotNull()).count()
rescued_pct = (rescued_records / total_records * 100) if total_records > 0 else 0

QUALITY_THRESHOLD_RESCUED = 1.0  # Alert if > 1% in rescued_data

print(f"\n📊 Results:")
print(f"   Total records: {total_records:,}")
print(f"   Rescued records: {rescued_records:,}")
print(f"   Rescued percentage: {rescued_pct:.4f}%")

if rescued_pct > QUALITY_THRESHOLD_RESCUED:
    print(f"\n🚨 QUALITY ALERT: {rescued_pct:.2f}% in _rescued_data (threshold: {QUALITY_THRESHOLD_RESCUED}%)")
    print("   Action: Investigate source format changes")
    print("   Action: Review _rescued_data samples")
    
    # Show sample of rescued data
    print("\n🔍 Sample rescued records:")
    df.filter(col("_rescued_data").isNotNull()) \
        .select("device_id", "_rescued_data", "_source_file") \
        .show(5, truncate=False)
else:
    print(f"\n✅ PASS: Rescued data is {rescued_pct:.4f}% (threshold: {QUALITY_THRESHOLD_RESCUED}%)")

# ============================================================================
# QUALITY CHECK 2: Null Value Analysis
# ============================================================================
print("\n" + "=" * 80)
print("🔍 CHECK 2: NULL VALUE ANALYSIS")
print("=" * 80)

QUALITY_THRESHOLD_NULLS = 10.0  # Alert if > 10% nulls in critical columns

critical_columns = ["device_id", "user_id", "calories_burnt", "num_steps"]

print(f"\n📊 Checking critical columns: {critical_columns}\n")

null_results = []
for col_name in critical_columns:
    null_count = df.filter(col(col_name).isNull()).count()
    null_pct = (null_count / total_records * 100) if total_records > 0 else 0
    
    status = "✅ PASS" if null_pct <= QUALITY_THRESHOLD_NULLS else "🚨 FAIL"
    print(f"{status} {col_name:20s}: {null_count:,} nulls ({null_pct:.2f}%)")
    
    if null_pct > QUALITY_THRESHOLD_NULLS:
        print(f"      ⚠️  Exceeds threshold of {QUALITY_THRESHOLD_NULLS}%")
        null_results.append(col_name)

if null_results:
    print(f"\n🚨 QUALITY ALERT: High null rates in {len(null_results)} columns")
    print(f"   Columns: {', '.join(null_results)}")
else:
    print("\n✅ PASS: All critical columns within null threshold")

# ============================================================================
# QUALITY CHECK 3: Value Range Validation
# ============================================================================
print("\n" + "=" * 80)
print("🔍 CHECK 3: VALUE RANGE VALIDATION")
print("=" * 80)

print("\n📊 Validating numeric ranges...\n")

# Define valid ranges
range_checks = [
    ("calories_burnt", 0, 10000),
    ("num_steps", 0, 100000),
    ("miles_walked", 0, 100)
]

outlier_summary = []

for col_name, min_val, max_val in range_checks:
    outliers = df.filter(
        (col(col_name) < min_val) | (col(col_name) > max_val)
    ).count()
    
    outlier_pct = (outliers / total_records * 100) if total_records > 0 else 0
    
    status = "✅ PASS" if outliers == 0 else "⚠️  WARNING"
    print(f"{status} {col_name:20s}: {outliers:,} outliers ({outlier_pct:.2f}%)")
    print(f"      Valid range: [{min_val:,}, {max_val:,}]")
    
    if outliers > 0:
        outlier_summary.append((col_name, outliers))

if outlier_summary:
    print(f"\n⚠️  Found outliers in {len(outlier_summary)} columns")
    print("   Recommendation: Create quarantine table for investigation")
else:
    print("\n✅ PASS: All values within expected ranges")

# ============================================================================
# QUALITY CHECK 4: Referential Integrity
# ============================================================================
print("\n" + "=" * 80)
print("🔍 CHECK 4: REFERENTIAL INTEGRITY (Bronze → Silver)")
print("=" * 80)

silver_table = "poc_deepdive.silver.iot_telemetry_cleaned"

bronze_count = spark.table(bronze_table).count()
silver_count = spark.table(silver_table).count()

filtered_count = bronze_count - silver_count
filtered_pct = (filtered_count / bronze_count * 100) if bronze_count > 0 else 0

print(f"\n📊 Data Flow:")
print(f"   Bronze records: {bronze_count:,}")
print(f"   Silver records: {silver_count:,}")
print(f"   Filtered out: {filtered_count:,} ({filtered_pct:.2f}%)")

EXPECTED_FILTER_RATE = 5.0  # Expect < 5% to be filtered

if filtered_pct > EXPECTED_FILTER_RATE:
    print(f"\n⚠️  WARNING: High filter rate ({filtered_pct:.2f}%)")
    print(f"   Expected: < {EXPECTED_FILTER_RATE}%")
    print("   Action: Review silver layer filter conditions")
    print("   Action: Check source data quality")
else:
    print(f"\n✅ PASS: Filter rate within expected range")

# ============================================================================
# QUALITY CHECK 5: Duplicate Detection
# ============================================================================
print("\n" + "=" * 80)
print("🔍 CHECK 5: DUPLICATE DETECTION")
print("=" * 80)

silver_df = spark.table(silver_table)

# Check for duplicates by device_id + timestamp
duplicates = silver_df.groupBy("device_id", "timestamp_parsed") \
    .agg(count("*").alias("count")) \
    .filter(col("count") > 1)

duplicate_count = duplicates.count()

print(f"\n📊 Duplicate Analysis (device_id + timestamp):")
print(f"   Unique combinations: {silver_df.count():,}")
print(f"   Duplicate keys: {duplicate_count:,}")

if duplicate_count > 0:
    print(f"\n🚨 QUALITY ALERT: Found {duplicate_count} duplicate keys in Silver!")
    print("   Action: Verify deduplication logic")
    print("\n🔍 Sample duplicates:")
    duplicates.orderBy(col("count").desc()).show(10)
else:
    print("\n✅ PASS: No duplicates detected")

print("\n" + "=" * 80)
print("✅ DATA QUALITY VALIDATION COMPLETE")
print("=" * 80)

print("""
📊 Quality Checks Performed:
   1. ✅ Rescued data monitoring (< 1%)
   2. ✅ Null value analysis (< 10% in critical columns)
   3. ✅ Value range validation (detect outliers)
   4. ✅ Referential integrity (bronze → silver)
   5. ✅ Duplicate detection

💡 Production Best Practices:
   • Run quality checks AFTER each layer ingestion
   • Alert on threshold breaches (don't wait for manual review)
   • Quarantine bad data instead of dropping silently
   • Track quality metrics over time (trending)
   • Include quality checks in CI/CD pipeline

🎤 Interview Answer:
"Every production pipeline needs automated quality gates. We check
rescued_data percentage, null rates, value ranges, and referential
integrity. If rescued_data exceeds 1%, we alert immediately. Bad records
go to quarantine tables for investigation. We've prevented multiple
production incidents by failing fast on quality issues."
""")
````

````python
# HANDS-ON EXERCISE 7: Backfill Patterns
# How to reprocess data without duplicates

print("=" * 80)
print("HANDS-ON 7: BACKFILL PATTERNS")
print("=" * 80)
print("\n🎯 Goal: Learn safe backfill strategies")

bronze_table = "poc_deepdive.bronze.iot_telemetry_complete"
silver_table = "poc_deepdive.silver.iot_telemetry_cleaned"
gold_table = "poc_deepdive.gold.iot_daily_summary"

# ============================================================================
# PATTERN 1: Targeted Date Range Backfill
# ============================================================================
print("\n" + "=" * 80)
print("🎯 PATTERN 1: TARGETED DATE RANGE BACKFILL")
print("=" * 80)

print("""
📚 Use Case: Bug fix in silver transformation for specific dates

Steps:
1. Identify the problematic date range
2. Delete affected data from silver (and gold if needed)
3. DO NOT touch bronze or checkpoints
4. Rerun silver pipeline - it processes only affected dates
""")

print("\n🔍 Current state before backfill:")
silver_count_before = spark.table(silver_table).count()
print(f"   Silver records: {silver_count_before:,}")

# Show date distribution
print("\n📅 Silver records by date:")
spark.sql(f'''
    SELECT date, COUNT(*) as records
    FROM {silver_table}
    GROUP BY date
    ORDER BY date
    LIMIT 10
''').show()

print('''
\n🛠️  Backfill Example (commented to avoid actual deletion):

# Step 1: Identify target date range
target_dates = "('2024-01-01', '2024-01-02', '2024-01-03')"

# Step 2: Delete from silver (affected dates only)
spark.sql(f"""
    DELETE FROM {silver_table}
    WHERE date IN {target_dates}
""")
print("Deleted records from silver for target dates")

# Step 3: Delete from gold (if needed)
spark.sql(f"""
    DELETE FROM {gold_table}
    WHERE date IN {target_dates}
""")
print("Deleted records from gold for target dates")

# Step 4: Rerun silver pipeline
# Bronze checkpoint is untouched, so only new data is processed
# But Delta time travel lets silver read the specific bronze dates

silver_stream = (spark.readStream
    .option("startingVersion", 0)  # Read from beginning
    .table(bronze_table)
    .filter(col("_load_date").isin(target_dates))
    # ... rest of transformations
)

# The checkpoint will skip files it already processed
# Only affected dates get reprocessed
''')

print("✅ Pattern 1 documented (no actual deletion performed)")

# ============================================================================
# PATTERN 2: Full Reprocessing
# ============================================================================
print("\n" + "=" * 80)
print("🎯 PATTERN 2: FULL REPROCESSING")
print("=" * 80)

print("""
📚 Use Case: Major logic change requires reprocessing all historical data

Steps:
1. BACKUP checkpoint directory (critical!)
2. Delete checkpoint
3. Drop and recreate target table
4. Rerun pipeline - processes all files from beginning

Risks:
- Expensive (processes all data)
- Takes time proportional to total data volume
- Can impact production if not done carefully

Best Practice:
- Run in separate workspace/environment
- Use separate checkpoint path (e.g., checkpoint_backfill_20240601)
- Compare results before switching over
""")

print('''
\n🛠️  Full Reprocessing Example (commented):

# Step 1: Backup checkpoint
import datetime
backup_suffix = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
original_checkpoint = "/Volumes/poc_deepdive/_infrastructure/checkpoints/silver_iot"
backup_checkpoint = f"{original_checkpoint}_backup_{backup_suffix}"

dbutils.fs.cp(original_checkpoint, backup_checkpoint, recurse=True)
print(f"Checkpoint backed up to: {backup_checkpoint}")

# Step 2: Delete checkpoint and table
dbutils.fs.rm(original_checkpoint, recurse=True)
spark.sql(f"DROP TABLE IF EXISTS {silver_table}")
print("Checkpoint and table deleted")

# Step 3: Rerun pipeline
# Uses same code as original pipeline
# Processes ALL files from bronze
silver_stream = (spark.readStream
    .table(bronze_table)
    # ... all transformations
)

query = (silver_stream.writeStream
    .format("delta")
    .option("checkpointLocation", original_checkpoint)  # Fresh checkpoint
    .trigger(availableNow=True)
    .toTable(silver_table)
)

query.awaitTermination()
print("Full reprocessing complete!")

# Step 4: Validate results
old_count = spark.read.table(f"{silver_table}_old").count()  # Backup table
new_count = spark.table(silver_table).count()
print(f"Old count: {old_count:,} | New count: {new_count:,}")
''')

print("✅ Pattern 2 documented (no actual reprocessing)")

# ============================================================================
# PATTERN 3: Incremental Backfill (Large Datasets)
# ============================================================================
print("\n" + "=" * 80)
print("🎯 PATTERN 3: INCREMENTAL BACKFILL (Large Datasets)")
print("=" * 80)

print("""
📚 Use Case: Reprocess 10TB of data, but can't afford one big job

Strategy: Process in controlled batches

Steps:
1. Create temporary checkpoint path
2. Use maxFilesPerTrigger or maxBytesPerTrigger
3. Run multiple smaller jobs instead of one large job
4. Monitor progress and cost
""")

print('''
\n🛠️  Incremental Backfill Example:

# Delete target table but keep bronze
spark.sql(f"DROP TABLE IF EXISTS {silver_table}")

# Create backfill checkpoint (separate from production)
backfill_checkpoint = "/Volumes/poc_deepdive/_infrastructure/checkpoints/silver_backfill"

# Configure incremental processing
silver_stream = (spark.readStream
    .table(bronze_table)
    # ... transformations
)

query = (silver_stream.writeStream
    .format("delta")
    .option("checkpointLocation", backfill_checkpoint)
    .option("cloudFiles.maxFilesPerTrigger", 1000)  # Process 1000 files per run
    .trigger(availableNow=True)  # Process one batch then stop
    .toTable(silver_table)
)

query.awaitTermination()

# Check progress
processed = spark.table(silver_table).count()
total_bronze = spark.table(bronze_table).count()
progress_pct = (processed / total_bronze * 100)
print(f"Progress: {processed:,} / {total_bronze:,} ({progress_pct:.1f}%)")

# Repeat this job daily/hourly until complete
# Each run processes next 1000 files
# Checkpoint ensures no duplicates
''')

print("✅ Pattern 3 documented")

# ============================================================================
# BEST PRACTICES
# ============================================================================
print("\n" + "=" * 80)
print("💡 BACKFILL BEST PRACTICES")
print("=" * 80)

print("""
1. **Always Backup Checkpoints**
   - Version checkpoints before any backfill
   - Keep backup for at least 30 days
   - Document checkpoint paths in runbook

2. **Use Separate Checkpoint Paths for Backfills**
   - Production: /checkpoints/silver_iot
   - Backfill: /checkpoints/silver_iot_backfill_20240601
   - Prevents contaminating production state

3. **Test in Dev/Staging First**
   - Clone production table to dev
   - Run backfill logic in dev
   - Validate results before production

4. **Monitor Costs**
   - Backfills can be expensive (reprocessing TBs)
   - Use smaller clusters for non-urgent backfills
   - Consider spot/preemptible instances

5. **Backfill Bronze, Not Source**
   - Don't delete and re-ingest from S3
   - Bronze is your source of truth
   - Use Delta time travel to read specific versions

6. **Cascade Backfills Carefully**
   - Bronze backfill → Silver auto-processes new data
   - Silver backfill → Gold auto-processes new data
   - Don't need to manually trigger downstream layers

7. **Document Everything**
   - Why: What bug was being fixed?
   - When: Start and end time of backfill
   - What: Which dates/files were reprocessed?
   - Who: Who approved and executed?
   - Results: Before/after record counts
""")

print("\n" + "=" * 80)
print("✅ BACKFILL PATTERNS COMPLETE")
print("=" * 80)

print("""
🎤 Interview Answer:
"For backfills, we use three patterns depending on scope. For targeted fixes
(specific dates), we delete affected data from silver/gold but keep bronze and
checkpoints. Delta time travel lets us read those specific bronze partitions.
For full reprocessing, we backup the checkpoint, delete it, and rerun from
beginning - expensive but sometimes necessary. For large datasets, we use
maxFilesPerTrigger to process in controlled batches, running the job multiple
times. Key rule: always backup checkpoints before any backfill, and use
separate checkpoint paths for backfill jobs to avoid contaminating production state."
""")
````
